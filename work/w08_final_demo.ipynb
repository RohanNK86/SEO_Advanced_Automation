# 🎤 Week 8 Showcase — 5 Minute Demo Outline

## Week 8 Showcase — AI-Powered SEO Content Intelligence Platform

### Demo Question (30 sec)

**Business Question**

How can FlyRank automatically identify webpages that are losing SEO performance and recommend evidence-backed content refreshes instead of manually auditing thousands of pages?

This is the content intelligence problem addressed throughout this capstone.

### Method (90 sec)

**End-to-End Pipeline**

1. Collect webpage performance metrics (CTR, impressions, rankings, engagement, search volume).
2. Engineer refresh-related features and semantic embeddings.
3. Train a machine learning classifier to predict refresh priority.
4. Cluster similar webpages using embedding similarity.
5. Store semantic vectors in Pinecone.
6. Retrieve relevant context with RAG.
7. Generate explainable refresh recommendations using NVIDIA Nemotron through LangChain.
8. Serve recommendations through FastAPI and React.

### One Key Visualization (60 sec)

**Show:** Refresh Priority Distribution or Feature Importance chart.

**Explain:**

- High-priority pages consistently share declining CTR and rankings.
- Semantic similarity groups pages by topical refresh opportunities.
- The visualization helps content teams prioritize work instead of reviewing every page.

### One Honest Result (60 sec)

**What Worked**

- ML successfully prioritizes pages requiring refresh.
- RAG grounds recommendations using retrieved SEO context.
- AI agent produces structured recommendations instead of generic suggestions.

**Limitation**

- Recommendations depend on historical SEO signals and retrieved context.
- Human editors remain responsible for final publishing decisions.

### Recommendation (60 sec)

Deploy the platform as an internal content operations assistant.

Content teams receive:

- High-priority refresh queue.
- Evidence-backed SEO recommendations.
- Similar content references.
- Explainable reasoning before publishing updates.

This reduces manual auditing effort while improving consistency across large content libraries.
